## Cómo construyo mi inventario

Busco desde la noticia de Espinof, sigo entrevistas y guías y anoto también hipótesis concretas. La selección no es aleatoria ni exhaustiva. Registro una obra o familia por fila; dos escenas de Pulp Fiction cuentan una sola vez. No sumo cada objeto de Dexter. No mezclo las películas de un actor con referencias del juego.

Distingo **declarada**, cuando un creador reconoce la influencia (incluidos cuatro referentes generales del estudio); **reportada**, cuando una página describe un paralelismo concreto; y **pista**, cuando la conexión queda débil, ambigua o pendiente de concretar. Son clases editoriales de relación, no puntuaciones de veracidad. Una obra puede tener influencia declarada y un guiño visual sin verificar; conservo esa diferencia en su límite.

En Arma letal y Bad Boys identifico la saga a partir de los personajes nombrados. No asigno año ni adaptación a franquicias ambiguas. En la referencia a Hulk las fuentes apuntan a películas diferentes: registro la franquicia y excluyo asignar una entrega concreta.

```python
from pathlib import Path
import json, csv

# localizo mi carpeta al abrir el cuaderno desde ella
base = Path.cwd()
registros = json.loads((base / 'registro_manual.json').read_text(encoding='utf-8'))
fuentes = json.loads((base / 'fuentes.json').read_text(encoding='utf-8'))
ids_fuentes = {f['id'] for f in fuentes}
assert len({r['id'] for r in registros}) == len(registros)
assert len({r['titulo'].casefold() for r in registros}) == len(registros)
for registro in registros:
    assert set(registro['fuentes']) <= ids_fuentes
    assert registro['estado'] in {'declarada', 'reportada', 'pista'}
    assert registro['detalle'] and registro['limite']
with (base / 'referencias.csv').open('w', newline='', encoding='utf-8') as archivo:
    tabla = csv.DictWriter(archivo, fieldnames=list(registros[0]))
    tabla.writeheader()
    for registro in registros:
        tabla.writerow({**registro, 'fuentes': ';'.join(registro['fuentes'])})
(base / 'referencias.json').write_text(json.dumps(registros, ensure_ascii=False, indent=2), encoding='utf-8')
print(f'He preparado {len(registros)} fichas y {len(fuentes)} fuentes.')
```
