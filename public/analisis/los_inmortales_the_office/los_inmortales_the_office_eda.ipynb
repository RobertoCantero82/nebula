## 2. Preparación de mis datos

Cargo mi tabla de hechos con un campo por fila. Mantengo separadas la transcripción de pantalla y las fichas institucionales.

```python
# cargo mi tabla local sin dependencias externas
from pathlib import Path
import csv, unicodedata
with Path('los_inmortales_the_office_datos.csv').open(encoding='utf-8', newline='') as archivo:
    datos = list(csv.DictReader(archivo))
assert len({(f['obra'],f['campo']) for f in datos}) == len(datos)
assert all(f['valor'] and f['fuente'] for f in datos)
print(f'Registros documentales: {len(datos)}')
```

## 3. Comparación

Calculo la distancia entre los años y comparo las ciudades que ya he transcrito. Esa igualdad no descubre automáticamente un vínculo causal; reproduce el cruce documental que origina el artículo.

```python
# comparo los años y las ciudades documentadas
def valor(obra, campo):
    return next(f['valor'] for f in datos if f['obra']==obra and f['campo']==campo)
def normalizo(texto):
    return unicodedata.normalize('NFKC', texto).strip().casefold()
distancia = int(valor('The Office (EE. UU.)','anio')) - int(valor('Los inmortales','anio'))
ciudad_brenda = valor('Los inmortales','ciudad_nacimiento_personaje')
ciudad_oficina = valor('The Office (EE. UU.)','ciudad_sucursal')
assert distancia == 19
assert normalizo(ciudad_brenda) == normalizo(ciudad_oficina)
print(f'Distancia entre años de estreno: {distancia}')
print(f'Ciudad de nacimiento de Brenda: {ciudad_brenda}')
print(f'Ciudad de la sucursal: {ciudad_oficina}')
print('Transcripción:', valor('Los inmortales','transcripcion_legible'))
```
