## Cómo preparo los datos

Conservo el texto de evidencia y las referencias; no completo con mi memoria los casos sin prueba. Compruebo identificadores, categorías y años antes de generar la tabla de análisis.

```python
# localizo mis archivos sin necesitar acceso a internet
from pathlib import Path
import csv
import json
raiz = Path.cwd()
with (raiz / 'data/raw/peliculas_codificadas.csv').open(encoding='utf-8', newline='') as archivo:
    peliculas = list(csv.DictReader(archivo))
print('Películas documentadas:', len(peliculas))
```

Leo 25 registros. La tabla es pequeña porque cada película aporta una sola observación, aunque Bond pueda ser capturado varias veces dentro de ella. No interpreto 25 filas como 25 misiones comparables.

```python
# compruebo que mi tabla distingue hechos positivos de casos desconocidos
assert len(peliculas) == 25
assert len({p['id'] for p in peliculas}) == 25
assert len({p['actor'] for p in peliculas}) == 6
assert [int(p['anio']) for p in peliculas] == sorted(int(p['anio']) for p in peliculas)
assert {p['tipo'] for p in peliculas} <= {'captura', 'encierro_trampa', 'no_documentada'}
assert all(p['evidencia'] and p['fuente_url'].startswith('https://') for p in peliculas)
assert all(p['acceso_voluntario'] in {'si', 'no'} for p in peliculas)
print('Compruebo 25 identificadores únicos, seis intérpretes y evidencia enlazada en cada registro.')
```

## Mi decisión de limpieza

No sustituyo `no_documentada` por cero. Añado indicadores 1 para evidencia positiva y vacío para lo demás. Un vacío conserva la incertidumbre y no se usa para concluir que no hubo capturas.

```python
# separo la codificación documental de las variables que utilizaré en el análisis
for pelicula in peliculas:
    pelicula['captura_documentada'] = '1' if pelicula['tipo'] == 'captura' else ''
    pelicula['criterio_conservador'] = '1' if pelicula['tipo'] == 'captura' and pelicula['acceso_voluntario'] == 'no' else ''
    pelicula['criterio_ampliado'] = '1' if pelicula['tipo'] in {'captura', 'encierro_trampa'} else ''
salida = raiz / 'data/processed/peliculas_analisis.csv'
with salida.open('w', encoding='utf-8', newline='') as archivo:
    escritor = csv.DictWriter(archivo, fieldnames=list(peliculas[0]))
    escritor.writeheader()
    escritor.writerows(peliculas)
print('Guardo una tabla de análisis con', len(peliculas), 'filas.')
```
