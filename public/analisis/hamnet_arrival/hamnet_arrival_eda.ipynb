## 1. Fuentes, glosario y límites

Llamo partitura original a la música compuesta para una película, y pieza preexistente a una obra anterior incorporada a ella. Distingo los créditos de las interpretaciones narrativas. Reúno un pequeño conjunto manual de hechos, un campo por fila, con su fuente y tipo de evidencia. Consulto las fuentes el 19 de noviembre de 2026.

No utilizo MovieLens 32M: su corte temporal en 2023 no cubre Hamnet (2025). No invento notas, etiquetas ni espectadores comunes. Los años corresponden a los estrenos originales y no a los españoles. La ficha de AFI combina autor del relato y guionista bajo Writer: separo sus funciones con ese alcance y no cuento a Chiang como guionista de la adaptación.

```python
# cargo mi tabla local para reproducir el análisis sin conexión
from pathlib import Path
import csv, json
ruta = Path('hamnet_arrival_datos.csv')
with ruta.open(encoding='utf-8', newline='') as archivo:
    datos = list(csv.DictReader(archivo))
print(f'Registros: {len(datos)}')
for fila in datos:
    assert fila['fuente'].startswith('https://')
    assert fila['valor']
assert len({(f['pelicula'], f['campo']) for f in datos}) == len(datos)
print('Compruebo campos, fuentes y claves únicas.')
```

## 2. Preparación

Organizo cada película por campos. No intento cruzar todos los créditos: comparo únicamente dirección, guion, origen literario, partitura y la pieza compartida. Conservo los nombres tal como los he normalizado en la tabla. La coincidencia de dos títulos musicales exige además comprobar autoría y contexto en las fuentes.

```python
# preparo dos fichas y separo los datos de la obra musical
fichas = {nombre: {f['campo']: f['valor'] for f in datos if f['pelicula'] == nombre} for nombre in ['Arrival', 'Hamnet']}
print(json.dumps(fichas, ensure_ascii=False, indent=2))
```

## 3. Análisis: distancia y diferencias

Calculo la diferencia entre años. Después comparo los campos seleccionados; una diferencia de nombre no demuestra ausencia total de profesionales compartidos.

```python
# comparo años y campos concretos, sin extenderlos al equipo completo
diferencia = int(fichas['Hamnet']['anio']) - int(fichas['Arrival']['anio'])
print(f'Diferencia entre años de estreno: {diferencia}')
for campo in ['direccion', 'guion', 'partitura_original']:
    print(campo, 'coincide:', fichas['Arrival'][campo] == fichas['Hamnet'][campo])
assert diferencia == 9
```

## 4. La conexión documentada

Compruebo que la misma obra aparece en ambas fichas. Sheppard, participante en la grabación utilizada en Arrival, confirma su presencia. El sello publica la lista oficial de Hamnet y Richter explica su uso en una entrevista con The Credits. Las fuentes prueban inclusión y decisión creativa; no equivalencia exacta de grabación, edición o duración entre películas.

```python
# verifico la coincidencia y la antigüedad de la pieza
obra = fichas['Arrival']['pieza_compartida']
assert obra == fichas['Hamnet']['pieza_compartida']
anio_obra = int(next(f['valor'] for f in datos if f['pelicula'] == 'Obra'))
print('Obra compartida:', obra)
print('Autor: Max Richter')
print('Publicación original:', anio_obra)
for nombre, ficha in fichas.items():
    print(nombre, 'se estrena', int(ficha['anio'])-anio_obra, 'años después del año de publicación de la obra.')
assert anio_obra < min(int(f['anio']) for f in fichas.values())
```
